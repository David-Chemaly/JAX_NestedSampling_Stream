```python
import corner
import numpy as np
import matplotlib.pyplot as plt
```

```python
size  = 100000
q     = np.random.uniform(0.5, 1.5, size)
x_hat = np.random.normal(0, 1, size)
y_hat = np.random.normal(0, 1, size)
z_hat = np.abs(np.random.normal(0, 1, size))

norm = np.sqrt(x_hat**2 + y_hat**2 + z_hat**2)
x_hat /= norm
y_hat /= norm
z_hat /= norm

samples = np.column_stack((q, x_hat, y_hat, z_hat))
figure = corner.corner(samples, 
              labels=['q', 'x_hat', 'y_hat', 'z_hat'], 
              color='blue',
              quantiles=[0.16, 0.5, 0.84], 
              show_titles=True, 
              title_kwargs={"fontsize": 12})
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x_hat, y_hat, z_hat, s=1, alpha=0.1)
ax.set_xlabel('x_hat')
ax.set_ylabel('y_hat')
ax.set_zlabel('z_hat')
plt.show()
```
